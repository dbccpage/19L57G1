# Falsify NSA-000\n\nAuthorityEffect: None. Deliberate breaks. Expected exceptions are the point.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path('..').resolve() / 'src'))
from nsa000.selection import evaluate_set, report_one
from nsa000.born_lift import born_from_weights, optional_amplitude_lift
```

Empty evaluation must fail.

```python
try:
    evaluate_set(frozenset())
    raise SystemExit('should have failed')
except ValueError as e:
    print('caught', e)
```

A pick outside the set must fail.

```python
try:
    report_one(frozenset({1, 2}), 9, declare=True)
    raise SystemExit('should have failed')
except ValueError as e:
    print('caught', e)
```

A silent singleton from a three-way tie is unauthorized.

```python
r = report_one(frozenset({1, 2, 3}), 1, declare=False)
assert r.unauthorized
print('unauthorized silent pick from', r.pool_size)
```

Negative weights must fail. Phase length mismatch must fail.

```python
for call in (
    lambda: born_from_weights((-1.0, 1.0)),
    lambda: optional_amplitude_lift((1.0, 1.0), (0.0,)),
):
    try:
        call()
        raise SystemExit('should have failed')
    except ValueError as e:
        print('caught', e)
```
