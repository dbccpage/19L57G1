## Sandcastles\nSame two objects. Two measures. Opposite winners.

```python
a = Castle('A', height=3, width=1)
b = Castle('B', height=1, width=3)
print('height elects', better(a, b, height_measure))
print('width elects', better(a, b, width_measure))
```

Move the sliders if ipywidgets is available. Otherwise the cell above is the whole point.

```python
try:
    import ipywidgets as w
    from IPython.display import display
    ha = w.FloatSlider(value=3, min=0, max=5, description='A height')
    wa = w.FloatSlider(value=1, min=0, max=5, description='A width')
    hb = w.FloatSlider(value=1, min=0, max=5, description='B height')
    wb = w.FloatSlider(value=3, min=0, max=5, description='B width')
    out = w.Output()
    def _go(_=None):
        out.clear_output()
        aa = Castle('A', ha.value, wa.value)
        bb = Castle('B', hb.value, wb.value)
        with out:
            print('height', better(aa, bb, height_measure), 'width', better(aa, bb, width_measure))
    for s in (ha, wa, hb, wb):
        s.observe(_go, 'value')
    display(ha, wa, hb, wb, out)
    _go()
except Exception as e:
    print('widgets unavailable:', type(e).__name__)
```
